```python
import pandas as pd
import joblib
from sklearn.metrics import classification_report, confusion_matrix

# 1. Load your test data and the saved model
X_test = pd.read_csv('../data/processed/X_test.csv')
y_test = pd.read_csv('../data/processed/y_test.csv').values.ravel()
model = joblib.load('../models/predictive_maintenance_model.pkl')

# 2. IMPORTANT: Apply the same cleaning/encoding to test data
# Drop identifiers
cols_to_drop = ['Product ID', 'UDI']
X_test = X_test.drop(columns=[c for c in cols_to_drop if c in X_test.columns], errors='ignore')
# Encode 'Type' (must use the same method as training)
X_test = pd.get_dummies(X_test, columns=['Type'], drop_first=True)

# 3. Make predictions
y_pred = model.predict(X_test)

# 4. Generate report
print("Confusion Matrix:")
print(confusion_matrix(y_test, y_pred))
print("\nClassification Report:")
print(classification_report(y_test, y_pred))#Of all the machines that actually failed, how many did the model correctly catch?
```

```python
# 1. Load the training data
X_train = pd.read_csv('../data/processed/X_train.csv')
y_train = pd.read_csv('../data/processed/y_train.csv').values.ravel()

# 2. Apply the SAME cleaning/encoding steps used for Test data
# Drop identifiers
cols_to_drop = ['Product ID', 'UDI']
X_train = X_train.drop(columns=[c for c in cols_to_drop if c in X_train.columns], errors='ignore')

# Encode 'Type' (using the same columns as the model was trained on)
X_train = pd.get_dummies(X_train, columns=['Type'], drop_first=True)

# 3. Make predictions on training data
y_train_pred = model.predict(X_train)

# 4. Generate report for Training Data
print("--- Training Data Evaluation ---")
print("Confusion Matrix:")
print(confusion_matrix(y_train, y_train_pred))
print("\nClassification Report:")
print(classification_report(y_train, y_train_pred))
```
