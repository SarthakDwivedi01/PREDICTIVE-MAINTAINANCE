```python
import pandas as pd
import joblib
from sklearn.ensemble import RandomForestClassifier

# 1. Load the files
X_train = pd.read_csv('../data/processed/X_train.csv')
y_train = pd.read_csv('../data/processed/y_train.csv').values.ravel()

# 2. DROP non-predictive identifiers
cols_to_drop = ['Product ID', 'UDI'] 
X_train = X_train.drop(columns=[c for c in cols_to_drop if c in X_train.columns], errors='ignore')

# 3. CONVERT categorical 'Type' to numeric
# This creates new columns: Type_L, Type_M, Type_H (with 0s and 1s)
X_train = pd.get_dummies(X_train, columns=['Type'], drop_first=True)

# 4. Train the model
model = RandomForestClassifier(n_estimators=100, random_state=42, class_weight='balanced')
model.fit(X_train, y_train)

# 5. Save the model
joblib.dump(model, '../models/predictive_maintenance_model.pkl')
print("Model trained successfully!")
```
